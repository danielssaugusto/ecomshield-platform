# Baseline de classificação de intenção — Bitext

Avaliação reproduzível do baseline TF-IDF + Regressão Logística. O teste é isolado; esta análise não mede desempenho em PT-BR ou em produção.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
RESULT = ROOT / 'data/processed/bitext_intent_evaluation.json'
assert RESULT.exists(), 'Execute primeiro: python scripts/train_bitext_intent_baseline.py'
metrics = json.loads(RESULT.read_text())
pd.Series({key: metrics[key] for key in ('train_rows', 'test_rows', 'accuracy', 'macro_f1', 'weighted_f1')}, name='resultado')
```

## Interpretação responsável

A métrica alta é válida somente dentro da distribuição do Bitext, que é em inglês e híbrida/sintética. Não deve ser apresentada como precisão em reclamações brasileiras reais. Antes de integrar qualquer modelo a produto, é necessária validação externa com dados reais e rotulados.
