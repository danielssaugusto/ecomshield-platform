# TP4 — Classificador de intenções

Pipeline reproduzível TF-IDF + Regressão Logística com GridSearchCV, baseline de classe mais frequente, métricas por classe e ROC-AUC multiclasse.

```python
from pathlib import Path
import json
import subprocess
import pandas as pd

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
subprocess.run(['python3', 'scripts/train_intent_gridsearch.py'], cwd=ROOT, check=True)
RESULT = ROOT / 'data/processed/tp4_intent_evaluation.json'
metrics = json.loads(RESULT.read_text())
pd.DataFrame([
    {'modelo': 'Baseline: classe mais frequente', **metrics['baseline']},
    {'modelo': 'TF-IDF + LogisticRegression ajustado', **{k: metrics['tuned_model'][k] for k in ('accuracy', 'macro_f1', 'roc_auc_ovr_macro')}}
])
```

```python
pd.DataFrame(metrics['tuned_model']['per_class']).T.sort_index()[['precision', 'recall', 'f1', 'support']]
```
